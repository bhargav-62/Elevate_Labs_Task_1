```python
import pandas as pd

# Load the Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Display first and last records (Hint 1)
print("First 5 records:")
display(df.head())
print("\nLast 5 records:")
display(df.tail())

# Check data structure and stats (Hint 3)
print("\n--- Data Information ---")
df.info()
print("\n--- Statistical Summary ---")
display(df.describe())

# Check for missing values (Hint 7)
print("\n--- Missing Values ---")
print(df.isnull().sum())
```

Task 1: Exploratory Data Analysis Observations
1. Feature Classification:

Numerical: Age, Fare, SibSp, Parch

Categorical: Sex, Embarked

Ordinal: Pclass (1st, 2nd, 3rd class)

Binary: Survived (0 or 1)

2. Target Variable: Survived

3. Data Quality Issues: > * Missing Values: Found 177 in Age, 687 in Cabin, and 2 in Embarked.

Data Size: 891 entries make it suitable for initial machine learning modeling.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visual style
sns.set(style="whitegrid")

# Create a figure with multiple subplots
plt.figure(figsize=(15, 10))

# 1. Visualize Survival Count (Target Variable)
plt.subplot(2, 2, 1)
sns.countplot(x='Survived', data=df, palette='viridis')
plt.title('Distribution of Survival (Target Variable)')

# 2. Visualize Passenger Class (Ordinal Feature)
plt.subplot(2, 2, 2)
sns.countplot(x='Pclass', data=df, palette='magma')
plt.title('Count of Passengers by Class')

# 3. Visualize Age Distribution (Numerical Feature)
plt.subplot(2, 2, 3)
sns.histplot(df['Age'].dropna(), kde=True, color='blue')
plt.title('Age Distribution of Passengers')

# 4. Visualize Survival by Sex (Relationship)
plt.subplot(2, 2, 4)
sns.countplot(x='Survived', hue='Sex', data=df, palette='pastel')
plt.title('Survival Rate by Gender')

plt.tight_layout()
plt.show()
```
